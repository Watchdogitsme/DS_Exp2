# Experiment 2: Data Handling Techniques with NumPy and Pandas

**Aim:** To perform data loading, inspection, cleaning, transformation, aggregation, feature engineering, and relational operations using NumPy and Pandas on Indian weather and electricity demand data.


### 1. Dataset Loading and Initial Overview


```python
import numpy as np
import pandas as pd

# Load dataset
df = pd.read_csv("C:\\VS\\DS\\final_dataset.csv")

# Basic dimensions and structure
print("First 5 records:")
display(df.head())
print(f"Dataset Shape (Rows, Columns): {df.shape}")
print(f"Column Names: {list(df.columns)}")
```

**Inference:** The dataset consists of **105,980 rows** across **13 attributes** representing historical daily weather metrics (`tmax`, `tmin`, `humidity`, `heat_index`) and grid electricity demand figures (`peak_met_mw`, `peak_unmet_mw`, `total_demand_mw`) across Indian states.


### 2. Dataset Information and Statistical Description


```python
print("--- Dataset Information ---")
df.info()

print("\n--- Summary Statistics (Numerical Features) ---")
display(df.describe())
```

**Inference:** There are 9 continuous floating-point variables, 2 integer indicator variables (`hot_day`, `hot_night`), and 2 categorical/string variables (`date`, `state`). Maximum temperature spans from **-3.46°C** to **48.77°C**, while total power demand spans from **0 MW** to **30,982 MW** with a mean of **5,721.25 MW**.


### 3. Null Value Identification


```python
# Check for null values and their distribution across columns
null_counts = df.isnull().sum()
total_nulls = null_counts.sum()

print(f"Total Null Values: {total_nulls}\n")
print("Null count per column:")
print(null_counts[null_counts > 0])
```

**Inference:** Missing values are confined to the power demand variables and calculated temperature metrics (`peak_met_mw`, `peak_unmet_mw`, `tavg`, `temp_range`, `total_demand_mw`), accounting for **860 missing entries** (~0.16% of total rows).


### 4. Missing Value Imputation (inplace=True)


```python
# Impute missing values with column median using inplace=True
columns_to_impute = ['peak_met_mw', 'peak_unmet_mw', 'tavg', 'temp_range', 'total_demand_mw']
medians = df[columns_to_impute].median()

df[columns_to_impute] = df[columns_to_impute].fillna(medians)

print("Remaining Null Values after Imputation:")
print(df.isnull().sum().sum())
```

**Inference:** Median imputation preserves the original data distribution without being skewed by extreme demand surges or outlier values, resulting in zero remaining missing cells across the DataFrame.


### 5. Top Records Sorting and Conditional Filtering


```python
# Top 7 records with highest peak power demand met
print("Top 7 Highest Peak Met Demand Records:")
display(df.sort_values(by='peak_met_mw', ascending=False).head(7)[['date', 'state', 'tmax', 'peak_met_mw', 'total_demand_mw']])

# Top 7 records with highest heat index
print("\nTop 7 Highest Heat Index Records:")
display(df.sort_values(by='heat_index', ascending=False).head(7)[['date', 'state', 'tmax', 'humidity', 'heat_index', 'total_demand_mw']])
```

**Inference:** Maharashtra exhibits the highest power demand peaks in the country (>28,000 MW), typically coinciding with elevated summer temperatures (35°C–43°C).


### 6. Frequency Distribution (2 Cases)


```python
# Case 1: Frequency listing of categorical column 'state'
print("Case 1: Observation Count per State (Top 5):")
print(df['state'].value_counts().head(5))

# Case 2: Frequency listing of binary indicator 'hot_day'
print("\nCase 2: Frequency Distribution of 'hot_day':")
print(df['hot_day'].value_counts(normalize=True) * 100)
```

**Inference:** The dataset has an evenly distributed time series across all states (~2,922 days per state). The `hot_day` metric triggers on approximately **6.18%** of the observed days.


### 7. Implicit and Explicit Sorting of Rows and Columns


```python
# 1. Sorting rows by index (Implicit Indexing sort)
df_sorted_idx = df.sort_index(ascending=False)

# 2. Sorting rows explicitly by multiple column values
df_sorted_values = df.sort_values(by=['state', 'total_demand_mw'], ascending=[True, False])

# 3. Sorting column headers alphabetically
df_sorted_cols = df.reindex(sorted(df.columns), axis=1)

print("Columns sorted alphabetically:")
print(df_sorted_cols.columns.tolist())
```

**Inference:** Sorting by values allows hierarchical ordering (e.g., peak demand within each state), while axis-level reindexing organizes attributes alphabetically to standardize reporting schemas.


### 8. Conditional Filtering with Compound Conditions (3 Cases)


```python
# Case 1: High temperature (> 40°C) and extreme peak demand (> 25000 MW)
cond1 = (df['tmax'] > 40) & (df['total_demand_mw'] > 25000)
res1 = df.loc[cond1, ['date', 'state', 'tmax', 'total_demand_mw']].head(3)
print("Case 1: Extreme Temperature (>40°C) & High Demand (>25k MW):")
display(res1)

# Case 2: Unmet demand occurred (> 500 MW) during hot days
cond2 = (df['hot_day'] == 1) & (df['peak_unmet_mw'] > 500)
res2 = df.loc[cond2, ['date', 'state', 'tmax', 'peak_unmet_mw']].head(3)
print("\nCase 2: Hot Day with High Unmet Demand (>500 MW):")
display(res2)

# Case 3: Sub-zero minimum temperature (< 0°C) with low humidity (< 60%)
cond3 = (df['tmin'] < 0) & (df['humidity'] < 60)
res3 = df.loc[cond3, ['date', 'state', 'tmin', 'humidity', 'total_demand_mw']].head(3)
print("\nCase 3: Sub-zero Temperatures & Dry Conditions:")
display(res3)
```

**Inference:** Compound masking isolates critical grid-stress events, showing that grid deficits (`peak_unmet_mw`) concentrate in April heatwaves, whereas winter sub-zero conditions are prominent in northern regions like Jammu & Kashmir.


### 9. Minimum and Maximum Value Analysis


```python
# Identify indices and values of extreme parameters
max_demand_row = df.loc[df['total_demand_mw'].idxmax()]
min_temp_row = df.loc[df['tmin'].idxmin()]
max_temp_row = df.loc[df['tmax'].idxmax()]

print(f"Maximum Total Demand: {max_demand_row['total_demand_mw']} MW (State: {max_demand_row['state']}, Date: {max_demand_row['date']})")
print(f"Lowest Recorded Temp: {min_temp_row['tmin']}°C (State: {min_temp_row['state']}, Date: {min_temp_row['date']})")
print(f"Highest Recorded Temp: {max_temp_row['tmax']}°C (State: {max_temp_row['state']}, Date: {max_temp_row['date']})")
```

**Inference:** `idxmax()` and `idxmin()` pinpoint exact geographical and temporal records of extreme climatic and operational thresholds across the nation.


### 10. GroupBy on Single and Multiple Columns (2 Cases)


```python
# Case 1: Group by single column 'state'
grouped_state = df.groupby('state')['total_demand_mw'].mean().sort_values(ascending=False).head(5)
print("Case 1: Top 5 States by Average Power Demand (MW):")
display(grouped_state)

# Case 2: Group by multiple columns ('state', 'hot_day')
grouped_multi = df.groupby(['state', 'hot_day'])[['tmax', 'total_demand_mw']].mean().head(6)
print("\nCase 2: Average Metrics Grouped by State and Hot Day Status:")
display(grouped_multi)
```

**Inference:** Industrialized states consistently maintain mean base loads exceeding 15,000 MW. Grouping across (`state`, `hot_day`) reveals that power demand increases significantly on days flagged as `hot_day == 1`.


### 11. Feature Engineering (Adding Derived Columns)


```python
# 1. Unmet Demand Percentage (%)
df['unmet_demand_pct'] = (df['peak_unmet_mw'] / (df['total_demand_mw'] + 1e-5)) * 100

# 2. Daily Temperature Fluctuation Ratio
df['temp_swing_ratio'] = df['temp_range'] / (df['tavg'].abs() + 1e-5)

print("Derived Feature Preview:")
display(df[['date', 'state', 'peak_unmet_mw', 'total_demand_mw', 'unmet_demand_pct', 'temp_swing_ratio']].head())
```

**Inference:** Computing normalized ratios provides scale-invariant features, enabling uniform comparison of power deficits across states with differing baseline grid capacities.


### 12. GroupBy with Multiple Aggregations (2 Cases)


```python
# Case 1: Multiple aggregations on power demand
agg_demand = df.groupby('state').agg(
    mean_demand=('total_demand_mw', 'mean'),
    max_demand=('total_demand_mw', 'max'),
    std_demand=('total_demand_mw', 'std'),
    total_unmet=('peak_unmet_mw', 'sum')
).sort_values(by='mean_demand', ascending=False).head(5)

print("Case 1: State-wise Electricity Demand Aggregations:")
display(agg_demand)

# Case 2: Aggregations on temperature and heat index
agg_weather = df.groupby('state').agg(
    avg_max_temp=('tmax', 'mean'),
    record_max_temp=('tmax', 'max'),
    avg_heat_index=('heat_index', 'mean')
).sort_values(by='record_max_temp', ascending=False).head(5)

print("\nCase 2: State-wise Thermal Aggregations:")
display(agg_weather)
```

**Inference:** Using named aggregations computes multi-variable summaries in a single step, showing that states with higher standard deviations (`std_demand`) experience larger seasonal load fluctuations.


### 13. Group Selection and Conditional Group Filtering


```python
# Selection 1: Extract a specific group by name using get_group()
maharashtra_group = df.groupby('state').get_group('MAHARASHTRA')
print(f"Maharashtra Records Count: {len(maharashtra_group)}")

# Selection 2: Filter groups having an overall mean demand exceeding 12,000 MW
high_demand_states = df.groupby('state').filter(lambda x: x['total_demand_mw'].mean() > 12000)
print(f"States with Mean Demand > 12,000 MW: {high_demand_states['state'].unique().tolist()}")
```

**Inference:** `.get_group()` retrieves subsets for localized inspection, while `.filter()` retains only the top energy-consuming state subsystems.


### 14. Correlation Analysis


```python
# Compute Pearson correlation matrix across numerical variables
corr_cols = ['tmax', 'humidity', 'heat_index', 'peak_met_mw', 'total_demand_mw']
corr_matrix = df[corr_cols].corr()

print("Pearson Correlation Matrix:")
display(corr_matrix)
```

**Inference:** `peak_met_mw` and `total_demand_mw` have a near-perfect positive correlation ($r \approx 0.999$), showing that grid operators successfully meet demand on most days. A positive correlation exists between maximum temperature (`tmax`) and demand ($r \approx 0.285$), reflecting air conditioning loads during hotter periods.


### 15. Data Transformation (Min-Max Normalization)


```python
# Min-Max Normalization: (X - X_min) / (X_max - X_min)
norm_cols = ['total_demand_mw', 'tmax', 'heat_index']

df_normalized = df[norm_cols].copy()
for col in norm_cols:
    df_normalized[col + '_norm'] = (df[col] - df[col].min()) / (df[col].max() - df[col].min())

print("Normalized Values (Scaled to [0, 1]):")
display(df_normalized[[col + '_norm' for col in norm_cols]].head())
```

**Inference:** Min-Max normalization rescales heterogeneous units (MW and °C) into a uniform range $[0, 1]$, preventing features with larger numeric scales from dominating distance-based machine learning models.


### 16. Joining, Merging, and Concatenation


```python
# 1. Concatenation: Row-wise splitting and vertical combining
subset_a = df.iloc[:3, :4]
subset_b = df.iloc[3:6, :4]
concatenated_df = pd.concat([subset_a, subset_b], axis=0)

# 2. Merging: Joining state region metadata
region_lookup = pd.DataFrame({
    'state': ['MAHARASHTRA', 'GUJARAT', 'TAMIL NADU', 'SIKKIM', 'RAJASTHAN'],
    'region': ['West', 'West', 'South', 'North-East', 'North']
})

merged_df = pd.merge(df[['date', 'state', 'total_demand_mw']].head(10), region_lookup, on='state', how='left')

print("Concatenation Result:")
display(concatenated_df)
print("\nMerge with Regional Metadata:")
display(merged_df)
```

**Inference:** `pd.concat` vertically combines continuous observations, while `pd.merge` acts like a SQL relational join to enrich the primary dataset with geographic region data.

### Q1. Explain the concept of "Broadcasting" in NumPy with the help of a suitable example.

NumPy **Broadcasting** describes how arithmetic operations are handled between arrays of different shapes. The smaller array is logically "broadcast" (stretched) across the larger array without making unneeded memory copies, following two rules:
1. Array dimensions are compared element-wise from right to left (trailing dimensions first).
2. Two dimensions are compatible if they are equal, or if one of them is 1.


```python
import numpy as np

# Example of NumPy Broadcasting
demand_matrix = np.array([[1200, 1500, 1800],
                          [2200, 2500, 2800]])  # Shape: (2, 3)
scale_factors = np.array([1.1, 1.2, 1.3])       # Shape: (3,) -> Broadcasts to (2, 3)

scaled_demand = demand_matrix * scale_factors
print("Broadcasted Result:\n", scaled_demand)
```

### Q2. Explain the difference between label-based indexing (`df.loc[]`) and integer-based indexing (`df.iloc[]`) in Pandas with the help of a suitable example.

| Indexer | Indexing Mode | Slicing Boundary Behavior | Key Use Case |
| :--- | :--- | :--- | :--- |
| **`df.loc[]`** | Explicit **Labels / Names** (Index labels, column names) | **Inclusive** of both start and stop labels | Filtering by date labels, state names, or boolean conditions |
| **`df.iloc[]`** | Implicit **Integer Positions** ($0$ to $n-1$) | **Exclusive** of the stop integer index (Python standard) | Positional extraction (e.g., first 5 rows, last 3 columns) |


```python
# loc vs iloc Demonstration
sample_df = pd.DataFrame({'temp': [32.5, 41.2, 28.0]}, index=['Day_A', 'Day_B', 'Day_C'])

# Label-based: includes 'Day_B'
print("Using loc (Label-based):")
display(sample_df.loc['Day_A':'Day_B', ['temp']])

# Integer-position: index 2 is excluded (retrieves indices 0 and 1)
print("\nUsing iloc (Integer-based):")
display(sample_df.iloc[0:2, [0]])
```
